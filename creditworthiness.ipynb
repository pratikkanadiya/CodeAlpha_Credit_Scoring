* Cbal: Checking account balance
* Cdur: Credit duration
* Chist: Credit history
* Cpur: Purpose of the loan
* Camt: Loan amount
* Sbal: Savings account balance
* Edur: Employment duration
* InRate: Installment rate
* MSG: Marital status and gender
* Oparties: Other parties involved
* Rdur: Residence duration
* Prop: Property type
* Age: Applicant's age
* inPlans: Existing plans
* Htype: Housing type
* NumCred: Number of existing credits
* JobType: Job type
* Ndepend: Number of dependents
* telephone: Presence of a telephone
* foreign: Foreign worker status
* creditScore: Credit score

**Data Pre-Processing**

```python
from sklearn.model_selection import train_test_split

x = df.drop('creditScore', axis = 1)
y = df['creditScore']

x_trainA, x_testA, y_trainA, y_testA = train_test_split(x, y, test_size = 0.3, random_state = 42)
```

* Dataset A (For Logistic Regression): Scale Everything
* Apply Log Transform to Camt and Cdur.
* Scale age, InRate, and NumCred using StandardScaler

```python
from sklearn.preprocessing import FunctionTransformer, StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer

log_transformer = FunctionTransformer(func=np.log1p, inverse_func=np.expm1)
scale = StandardScaler()
label = LabelEncoder()
ohe = OneHotEncoder(handle_unknown='ignore')


categorical_features_for_ohe = x_trainA.select_dtypes(include=['object']).columns.to_list()

transformer = [
    ('log_features', log_transformer, ['Camt', 'Cdur']),
    ('scale_features', scale, ['age', 'InRate', 'NumCred']),
    ('categorical_features', ohe, categorical_features_for_ohe)
]

processor = ColumnTransformer(
    transformers = transformer, remainder='passthrough'
)

x_trainA_scaled = processor.fit_transform(x_trainA)
x_testA_scaled = processor.transform(x_testA)


y_trainA_scaled = label.fit_transform(y_trainA)
y_testA_scaled = label.transform(y_testA)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

regressor = LogisticRegression()
regressor.fit(x_trainA_scaled, y_trainA_scaled)
y_pred_regressor = regressor.predict(x_testA_scaled)
print(accuracy_score(y_testA_scaled, y_pred_regressor))
print(classification_report(y_testA_scaled, y_pred_regressor))
print(confusion_matrix(y_testA_scaled, y_pred_regressor))
print(regressor.intercept_)
print(regressor.classes_)
```

```python
from sklearn.model_selection import GridSearchCV

param = [
    {
        'penalty': ['l1'],
        'C': [1, 2, 3, 4, 5],
        'solver': ['liblinear', 'saga'],
        'max_iter': [100, 1000, 2500, 5000]
    },
    {
        'penalty': ['l2', None],
        'C': [1, 2, 3, 4, 5],
        'solver': ['newton-cg', 'lbfgs', 'sag', 'saga'],
        'max_iter': [100, 1000, 2500, 5000]
    },
    {
        'penalty': ['elasticnet'],
        'C': [1, 2, 3, 4, 5],
        'solver': ['saga'],
        'l1_ratio': [0.5],
        'max_iter': [100, 1000, 2500, 5000]
    }
]

grid = GridSearchCV(estimator = regressor, param_grid = param, cv = 5, n_jobs=-1, verbose=1) # Added n_jobs and verbose for better performance and output

grid.fit(x_trainA_scaled, y_trainA_scaled)

best_regressor = grid.best_estimator_
y_pred = best_regressor.predict(x_testA_scaled)

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
print('Best parameters found:', grid.best_params_)
print('Accuracy after GridSearchCV:', accuracy_score(y_testA_scaled, y_pred))
print('Classification Report after GridSearchCV:')
print(classification_report(y_testA_scaled, y_pred))
print('Confusion Matrix after GridSearchCV:')
print(confusion_matrix(y_testA_scaled, y_pred))
print('Intercept of best regressor:', best_regressor.intercept_)
print('Classes of best regressor:', best_regressor.classes_)
```

```python
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier()

tree.fit(x_trainA_scaled, y_trainA_scaled)
y_pred_tree = tree.predict(x_testA_scaled)

print(accuracy_score(y_testA_scaled, y_pred_tree))
print(classification_report(y_testA_scaled, y_pred_tree))
print(confusion_matrix(y_testA_scaled, y_pred_tree))
```
